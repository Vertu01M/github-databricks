```python
import pandas as pd
import numpy as np

# Load data from Unity Catalog table and convert to pandas
bronze_spark = spark.table('dataeng.projet1.reviews_baby_dataset_csv')
bronze = bronze_spark.toPandas()

print(f"Loaded {len(bronze)} rows")
bronze.head()
```

```python
# Check for null values in each column
print("Null counts per column:")
print(bronze.isna().sum())

# Check for duplicates
print(f"\nTotal rows: {len(bronze)}")
print(f"Duplicate rows: {bronze.duplicated().sum()}")

# Show DataFrame info
print("\nDataFrame info:")
bronze.info()
```

```python
# Detect missing values
missing_values = bronze.isnull().sum()
# Check for duplicate rows
duplicates = bronze.duplicated().sum()
# Summarize basic statistics
summary_stats = bronze.describe(include='all')
```

```python
# Make a copy to avoid altering the original dataset
Silver = bronze.copy()

# Drop rows with missing values
Silver.dropna(inplace=True)

# Drop duplicate rows
Silver.drop_duplicates(inplace=True)

# Reset the index
Silver.reset_index(drop=True, inplace=True)

# Save the cleaned Silver dataset to the DbAcademy workspace folder
output_path = '/Workspace/Users/vertubouka@gmail.com/github-databricks/Generative AI project/DbAcademy/silver_table.csv'
Silver.to_csv(output_path, index=False)
print(f"Silver table saved to: {output_path}")
print(f"Rows: {len(Silver)}, Columns: {len(Silver.columns)}")

# Display the cleaned dataset
display(Silver.head())


```
